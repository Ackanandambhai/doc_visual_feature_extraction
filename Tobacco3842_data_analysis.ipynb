```python
from pathlib import Path
from PIL import Image
import pandas as pd

# --------------------------------------------------
# Change this path to your dataset location
# --------------------------------------------------
DATASET_ROOT = Path("/Users/anandkumar/doc_visual_feature_extraction/data/raw")

# Supported image extensions
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png",
    ".bmp", ".tif", ".tiff",
    ".gif", ".webp"
}

records = []

folder_counts = {}

for subset in ["train", "test", "split"]:

    folder = DATASET_ROOT / subset

    if not folder.exists():
        print(f"{subset} folder not found.")
        continue

    count = 0

    for image_path in folder.rglob("*"):

        if image_path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        try:
            with Image.open(image_path) as img:

                width, height = img.size

                dpi = img.info.get("dpi", ("Not Available", "Not Available"))

                mode = img.mode

                if mode == "RGB":
                    channels = 3
                elif mode == "RGBA":
                    channels = 4
                elif mode == "L":
                    channels = 1
                else:
                    channels = len(img.getbands())

                file_size_kb = image_path.stat().st_size / 1024

                records.append({
                    "Subset": subset,
                    "Filename": image_path.name,
                    "Format": img.format,
                    "Width(px)": width,
                    "Height(px)": height,
                    "Resolution": f"{width} x {height}",
                    "Mode": mode,
                    "Channels": channels,
                    "DPI": dpi,
                    "File Size (KB)": round(file_size_kb, 2),
                })

                count += 1

        except Exception as e:
            print(f"Could not read {image_path}: {e}")

    folder_counts[subset] = count

# --------------------------------------------------
# Convert to DataFrame
# --------------------------------------------------

df = pd.DataFrame(records)

print("\n========== IMAGE COUNTS ==========")

for key, value in folder_counts.items():
    print(f"{key:10s}: {value}")

print("----------------------------------")
print(f"Total     : {sum(folder_counts.values())}")

print("\n========== SAMPLE ==========")
print(df.head())

# Save detailed report
"""df.to_csv("tobacco3842_image_report.csv", index=False)

print("\nSaved report to:")
print("tobacco3842_image_report.csv")"""
```
